Q2(a,b,c,d)

```python
brands = ['Samsung', 'Apple', 'Xiaomi', 'Oppo', 'Others']
share  = [30, 25, 20, 15, 10]
scores = np.random.randn(500)
fig,ax=plt.subplots(1,2,figsize=(12,5))

explode=(0.1,0,0,0,0)
ax[0].pie(share,labels=brands,autopct='%1.1f%%',explode=explode,shadow=True)
ax[0].set_title('Market Share of Mobile Brands');
ax[1].hist(scores,bins=20)
ax[1].set_title('Distribution of Scores');
plt.tight_layout()
plt.savefig('market.png')
plt.show()
```

```python
import seaborn as sns
import plotly.express as px

iris=sns.load_dataset('iris')
sns.scatterplot(x='sepal_length',y='petal_length',hue='species',data=iris)
plt.show()
```
